```python
import pymongo
import pandas as pd
```

a)

```python
client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["bestbuy"]
collection = db["queries"]
```

```python
bestbuy = pd.read_csv('bestbuy.csv')
bestbuy.head()
```

```python
bestbuy.columns = ["user", "sku", "category", "query", "click_time", "query_time"]

# Convert DF to dictionary
bestbuy_dict = bestbuy.to_dict(orient="records")
bestbuy_dict[0:2]
```

```python
len(bestbuy_dict)
```

```python
# Insert into MongoDB
collection.insert_many(bestbuy_dict)
```

c)

```python
top_skus = collection.aggregate([
    {"$group": {"_id": "$sku", "count": {"$sum": 1}}},  # Group by SKU and count occurrences
    {"$sort": {"count": -1}},
    {"$limit": 5}
])
```

```python
print("Top 5 most frequently searched SKUs:")
for sku in top_skus:
    print(f"SKU: {sku['_id']}, Count: {sku['count']}")
```
