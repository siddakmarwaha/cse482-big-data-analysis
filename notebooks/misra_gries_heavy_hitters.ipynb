```python
# Misra-Gries algorithm with buffer size of 1000
def misra_gries_algorithm(file_path, buffer_size=1000):
    buffer = {}  # To store article frequencies

    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            parts = line.strip().split()  # Split line by spaces
            article = parts[1]  # Extract article title
            
            if article in buffer:
                buffer[article] += 1  # Increment if in buffer
            elif len(buffer) < buffer_size:
                buffer[article] = 1  # Add new article if space available
            else:
                # Decrement all counters and remove those with 0 count
                keys_to_delete = []
                for key in buffer:
                    buffer[key] -= 1
                    if buffer[key] == 0:
                        keys_to_delete.append(key)
                
                # Remove zero-count entries
                for key in keys_to_delete:
                    del buffer[key]

    return buffer
```

```python
# Run the algorithm
file_path = "wiki_edit2.txt"
result = misra_gries_algorithm(file_path)

print("Frequent Articles and Their Estimated Counts:")
for article, count in sorted(result.items(), key=lambda x: -x[1]):  # Sort by frequency
    print(f"{article}: {count}")
```
